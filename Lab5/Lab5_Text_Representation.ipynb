# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
# Task 1: Cosine Similarity
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sentences)

cosine_sim = cosine_similarity(tfidf_matrix)

cosine_df = pd.DataFrame(
    cosine_sim,
    index=[f"Sentence {i+1}" for i in range(len(sentences))],
    columns=[f"Sentence {i+1}" for i in range(len(sentences))]
)

cosine_df
````

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
# Task 2: TF-IDF
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(sentences)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=[f"Sentence {i+1}" for i in range(len(sentences))]
)

display(tfidf_df)

# Important words in each sentence = words with the highest TF-IDF scores
for i, row in tfidf_df.iterrows():
    important_words = row[row > 0].sort_values(ascending=False)
    print(f"\n{i} important words:")
    print(important_words)
````

````python
# Task 3: Preprocess the Simpsons dataset and create Skip-Gram Word2Vec
import os
import pandas as pd
import numpy as np
import re
import nltk
import gensim
from nltk.tokenize import word_tokenize

# Download tokenizer resources if needed
nltk.download('punkt', quiet=True)
try:
    nltk.download('punkt_tab', quiet=True)
except:
    pass

# Load the provided Simpsons dataset.
# This works when the CSV is in the same folder as the notebook (e.g., after uploading to Colab).
csv_path = 'simpsons_script_lines.csv'
if not os.path.exists(csv_path):
    # Common Colab location
    csv_path = '/content/simpsons_script_lines.csv'

df = pd.read_csv(csv_path, low_memory=False)

# Use only the required column and remove missing values
spoken_words = df['spoken_words'].dropna().astype(str)

# Preprocessing procedure specified in the lab
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

cleaned_text = spoken_words.apply(clean_text)

# Tokenize each cleaned line
tokens = [word_tokenize(text) for text in cleaned_text]

# Remove any empty token lists
tokens = [token_list for token_list in tokens if len(token_list) > 0]

print("Number of tokenized lines:", len(tokens))
print("Example tokens:", tokens[0])

# Create the required Skip-Gram Word2Vec model
Skip_gram_model = gensim.models.Word2Vec(
    tokens,
    min_count=1,
    vector_size=100,
    window=5,
    sg=1,
    workers=1,
    seed=42
)

print("Vocabulary size:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
# Task 4: Find words similar to Homer, Marge, and Bart
for word in ['homer', 'marge', 'bart']:
    print(f"\nWords most similar to '{word}':")
    for similar_word, score in Skip_gram_model.wv.most_similar(word):
        print(f"{similar_word}: {score:.4f}")
````
